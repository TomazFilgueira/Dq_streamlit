# 2) Designing Interactive Sidebars

First, let's look at how we can separate information from the main app page using a sidebar. **Sidebars** in Streamlit allow us to add a layer of organization to our app by neatly separating different app features or categories, much like a table of contents. They can hold widgets, display additional information, and even serve as a navigation menu.

To create a sidebar in Streamlit, we use the `st.sidebar` function. To implement elements in a sidebar, we can use the sidebar's **object notation**, which follows the format `st.sidebar.[element]`. For example, `st.sidebar.write("Hello, world!")` adds text to the sidebar, just as `st.write("Hello, world!")` would add text to the main page. Here's an example:

```python
st.sidebar.title('My Sidebar')
st.sidebar.text('Hello, world!')
```



In this code, we're using `st.sidebar.title` and `st.sidebar.text` to add a title and text to the sidebar, similar to how `st.title` and `st.text` would be used for the main page. When we run this script, these elements will appear on the left side of our app, separate from the main page content.

Sidebars can hold most of the same elements as the main page. This includes text, images, buttons, sliders, and more. However, keep in mind that the sidebar has a narrower width, so some elements might appear differently.

One common use of sidebars is to control the content that appears on the main page. For example, we might use a sidebar selectbox to let users choose which content to display on the main page. Let's look at an example:

```python
option = st.sidebar.selectbox('Choose an option', ['Option 1', 'Option 2', 'Option 3'])

if option == 'Option 1':
    st.title('Option 1 Content')
    # Content for Option 1 goes here
elif option == 'Option 2':
    st.title('Option 2 Content')
    # Content for Option 2 goes here
else:
    st.title('Option 3 Content')
    # Content for Option 3 goes here
```



In this example, we're using a selectbox in the sidebar to let the user choose between three options. The content on the main page changes based on this selection, making our app more interactive and user-friendly.

Note that when the user makes a selection in the sidebar, the entire script reruns, just like when the user interacts with a widget on the main page. This is thanks to Streamlit's always up-to-date state, which we discussed in the previous lesson. This feature ensures that the sidebar and main page stay in sync, providing a seamless user experience.

# 3 Basic Columns

Now that we can navigate different parts of the GlobalJava Roasters app, we need to add employee information to the "About Us" section. With UI in mind, the cleanest way to do this is with **columns**.

**Columns** in Streamlit allow us to arrange widgets or display elements side by side, providing a more structured and organized layout. This can be handy in various scenarios, such as when we want to display related information or controls next to each other to make the interface more intuitive and user-friendly.

To create columns in Streamlit, we use the `st.columns` [function](https://docs.streamlit.io/develop/api-reference/layout/st.columns) and what's known as **multi-variable assignment**. Multi-variable assignment allows us to assign multiple variables at once. For instance, when we use the code `col1, col2 = st.columns(2)`, it creates two columns named `col1 and col1`. This works because function `st.columns(2)` returns a list of two columns, and `col1` and `col2` are assigned to these columns, respectively. This is considered best practice as it provides a clear and concise way to reference and manage each column. There are other methods to implement columns, but multi-variable assignment is generally the most straightforward and readable.

Here's an example of how we can use multi-variable assignment to write text in two columns:

```python
st.title('Main Page')

col1, col2 = st.columns(2)

col1.write('Hello')
col2.write('World')
```


This script places "Hello" and "World" side by side in the app. Columns can contain various elements like text, images, buttons, sliders, and more. Keep in mind that columns share the available width, so elements might display differently depending on whether they are on the main page or sidebar.

Columns of different widths can be created by passing a list of relative widths to st.columns. For example, `col1, col2 = st.columns([3, 1])` makes the first column [3] **three times wider** than the second [1]. Here's how this would affect our "Hello World" example (notice that "World" is now shifted further to the right).



Moreover, it's helpful to know that columns in Streamlit are essentially lists. This means they can be accessed and manipulated using indices. For example, `st.columns(2)[0]` would reference the first column, and `st.columns(2)[1]` would reference the second. This can be particularly useful in more dynamic or complex layouts.

# 4) 'With' Blocks

We've seen the basics of columns, but let's make our code more Pythonic by learning how to use **'with' statements** for more streamlined layout management in Streamlit.

The Python 'with' statement is a context manager, which you may have previously used to open files. A **context manager** in Python essentially sets up a temporary context for our code to run in and then cleans it up after the code is executed. This is especially useful in scenarios where you need to allocate and release resources precisely.

In Streamlit, 'with' blocks extend this concept of resource management to UI and layout elements, reducing code redundancy and improving readability. They start with the 'with' keyword, followed by a function call, and end with a colon. The code in the indented block that follows is executed within the context established by the 'with' statement. In Streamlit, this means that any setup or teardown associated with UI elements, like opening or closing layout sections, is automatically handled. For example, to create columns and add elements to them, we can use:

```python
col1, col2 = st.columns(2)
with col1:
    st.write('Hello')
with col2:
    st.write('World')
```

An added advantage of using 'with' blocks is the ability to reference Streamlit functions directly using `st`, instead of the specific column variable. This simplifies the code, making it more readable and concise. For instance, `st.write('Hello')` within a 'with' block automatically knows to write 'Hello' in the scope of the current block, be it `col1` or `col2`.

This code designates that 'Hello' is written in the first column and 'World' in the second. The use of 'with' blocks for each column also manages the scope and ensures clarity in our code.

## Nested 'with' Blocks

Nested 'with' blocks can also create sub-columns within a main column, offering a more complex layout. For instance, if we want to have a main column that itself contains two smaller columns, we can achieve this using nested 'with' blocks:

```python
main_col1, main_col2 = st.columns(2)

with main_col1:
    st.header("Main Column 1")

    sub_col1, sub_col2 = st.columns(2)
    with sub_col1:
        st.write("Sub-column 1")
    with sub_col2:
        st.write("Sub-column 2")

with main_col2:
    st.header("Main Column 2")

    sub_col3, sub_col4 = st.columns(2)
    with sub_col3:
        st.write("Sub-column A")
    with sub_col4:
        st.write("Sub-column B")
```



This example creates two main columns (`main_col1` and `main_col2`), each containing two sub-columns. The 'with' blocks are used to clearly define the scope of each main column and its sub-columns, allowing for a structured and organized layout that reflects the nested nature of the content.

'With' blocks in Streamlit simplify and enhance code readability, especially for complex layouts. Let's practice using 'with' blocks to streamline our app's layout.

# 5) Expanders

On the previous screens, we learned how to use 'with' blocks and columns in Streamlit to organize our app's layout. However, as we continue to add more features and content to our app, we may find that the interface becomes cluttered. To avoid overwhelming the user with too much information at once, we can use a feature in Streamlit called expanders.

**Expanders** allow us to hide or show information based on user interaction. When an expander is closed, it takes up minimal space and offers a summary or title of what the expander contains. When a user clicks on the expander, it opens to reveal the detailed information inside. This dynamic organization of content keeps your app's interface clean and user-friendly while providing all necessary information to the user.

We can create an expander by using the `st.expander()` function. We can then add content to the expander using the same methods we would use for the main app. Here's an example:

```python
with st.expander("Click to expand"):
    st.write("Detailed information inside the expander.")
```

In this example, "Click to expand" is the title that's always visible, even when the expander is closed. When the user clicks on the title, the expander opens to reveal the text "Detailed information inside the expander."

Like columns, expanders can be used within 'with' blocks to create more complex layouts. For example, we could create an expander within a column:

```python
col1, col2 = st.columns(2)
with col1:
    with st.expander("Click to expand"):
        st.write("Detailed information inside the expander.")
```

In this code, we've created two columns; within the first column, we've added an expander.

Expanders are a handy tool for organizing our app's content in a user-friendly way. By using expanders, we can ensure that our app's interface remains clean and uncluttered, even as we add more features and content.

However, it's important to use expanders judiciously. Overusing expanders can make our app's interface confusing and difficult to navigate. It's best to use expanders when we have a **large amount of secondary or supplemental information that we don't want to display all at once.**

# 6) Encapsulating Layouts in Functions

As we continue enhancing our app, it's important to keep our code organized. One effective way to do this is by encapsulating layouts within functions. This practice not only helps keep our code tidy but it also increases its modularity and reusability.

As a quick review from previous lessons, a function is a block of reusable code that performs a specific task. Because of this, when we **encapsulate a Streamlit layout within a function**, we are creating a reusable piece of user interface, which we can call upon whenever we need to display that particular layout. This can significantly simplify our main application script and make our code easier to maintain and debug.

Let's take a look at an example. In our app code from the previous screens, we have a large section dedicated to the "About Us" layout:

```python
if page == 'About Us':
    st.header('About Us')
    col1, col2, col3 = st.columns(3)

    with col1:
        st.image('Elizabeth_Bennet.png')
        st.header('Elizabeth Bennet')
        st.write('Founder and CEO, Elizabeth is passionate about bringing customers flavorful and delicious coffee.')

    with col2:
        st.image('Charles_Bingley.png')
        st.header('Charles Bingley')
        st.write('Marketing Director and Social Media Expert, Charles helps the world know about our great new flavors!')

    with col3:
        st.image('Georgiana_Darcy.png')
        st.header('Georgiana Darcy')
        st.write('Georgiana is the creative genius behind the scenes!')
```

We could encapsulate this layout in a function like this:

```python
def render_about_us():
    st.header('About Us')
    col1, col2, col3 = st.columns(3)

    with col1:
        st.image('Elizabeth_Bennet.png')
        st.header('Elizabeth Bennet')
        st.write('Founder and CEO, Elizabeth is passionate about bringing customers flavorful and delicious coffee.')
    with col2:
        st.image('Charles_Bingley.png')
        st.header('Charles Bingley')
        st.write('Marketing Director and Social Media Expert, Charles helps the world know about our great new flavors!')

    with col3:
        st.image('Georgiana_Darcy.png')
        st.header('Georgiana Darcy')
        st.write('Georgiana is the creative genius behind the scenes!')
```

Now, whenever we need to collect feedback, we call `render_about_us()`, like this:

```python
if page == 'About Us':
    render_about_us()
```

This makes our main application script cleaner and easier to read. It also allows us to reuse the feedback collection layout in other parts of our app, or even in other apps, without having to write the same code again.

An important aspect of Streamlit makes this possible: the always up-to-date app state. In Streamlit, every user interaction, such as typing in a text box or clicking a button, triggers the entire script to rerun. This unique feature ensures that the app's state is consistently refreshed, aligning the interface with the latest user inputs. This means that in our `render_about_us` function, we don't need to explicitly return values since the feedback is directly integrated into the app's state upon each interaction. This simplifies state management and is often referred to as Streamlit's 'magic'. It's a crucial concept that influences how we structure our code and manage the app's interactions.

Remember, encapsulating layouts within functions is a powerful technique for improving code organization and reusability in Streamlit. It's a good practice to adopt as we continue to build more complex Streamlit apps.

# 7) Tabs

In our ongoing exploration of Streamlit's layout features, we've seen how columns and expanders can help us organize content effectively. Now, let's turn our attention to another powerful tool: **tabs**.

Tabs allow us to divide our application into logical sections that can be accessed and viewed independently, which can be particularly useful for creating structured, easy-to-navigate interfaces.

In Streamlit, creating a tabbed interface involves using the `st.tabs` function. This function creates a new tabbed container in our application and returns an object we can use to add content to the tabs. Here's a basic example of how we can use `st.tabs`:

```python
tab1, tab2, tab3 = st.tabs(["Tab 1", "Tab 2", "Tab 3"])
with tab1:
    st.write("Text for Tab One")
with tab2:
    st.write("Another tab. Here's Tab Two")
with tab3:
    st.write("The final tab. This is Tab Three")
```



In this example, we've created a tabbed interface with three tabs labeled "Tab 1", "Tab 2", and "Tab 3". We've then added some simple content to each tab using the `st.write` function. When we run this code in Streamlit, we'll see a set of tabs at the top of the screen, and we can click on each tab to view its content.

Like other layout elements in Streamlit, tabs can be nested within each other or within 'with' blocks to create more complex layouts. For example, we could create a set of tabs within a column or a set of tabs within an expander.

In this code, we've created two columns; within the first column, we've added a tabbed interface.

```python
col1, col2 = st.columns(2)
with col1:
    st.header('Column 1')
with col2:
    tab1, tab2, tab3 = st.tabs(["Tab 1", "Tab 2", "Tab 3"])
    with tab1:
        st.write("Text for Tab One")
    with tab2:
        st.write("Another tab. Here's Tab Two")
    with tab3:
        st.write("The final tab. This is Tab Three")
```

# 8) Choosing the Right Layout Element

As we've seen, Streamlit offers a variety of layout elements to enhance the user interface of our apps. Choosing the right element — **columns**, **sidebars**, **expanders**, or **tabs** — depends on the specific content and the user experience we want to create. Let's compare and contrast these elements to help us make informed decisions for our app designs.

| Element   | Description                                                               | Best Uses                                                                 | Example Use Cases                                                                                                               |
|-----------|---------------------------------------------------------------------------|---------------------------------------------------------------------------|---------------------------------------------------------------------------------------------------------------------------------|
| Columns   | For arranging widgets or display elements side by side.                   | Comparative displays, horizontal organization of related content.         | Displaying charts next to each other for comparison. Side-by-side image and text presentation.                                  |
| Sidebars  | A navigation menu or a control panel, separate from main content.         | App navigation, grouping of global interactive controls.                  | Navigation menus for different app sections. Control panels for filtering or adjusting app-wide settings.                       |
| Expanders | Elements that can expand/collapse to show or hide additional information. | Managing content density, hiding optional or detailed information.        | FAQ sections with detailed answers hidden under questions. Advanced settings or additional options that aren't frequently used. |
| Tabs      | For categorizing content into separate, clickable sections.               | Categorization of content, reducing overwhelming information on one page. | Different categories of FAQs (e.g., General, Technical). Segmenting data analysis results into different thematic tabs.         |


**Making the Right Choice**

When deciding which element to use, consider the nature of your content and the user experience. Columns and sidebars **structure the layout**, while expanders and tabs **manage content density** and categorization. Often, a combination of these elements can provide the most effective layout. Remember, the goal is to create an intuitive and user-friendly app that makes information accessible and easy to navigate.

````python
def feedback_input():
    if 'feedback_submitted' not in st.session_state:
        st.session_state.feedback_submitted = False
    
    st.write("Did you enjoy your coffee?")
    yes_button_clicked = st.button("Yes!")
    no_button_clicked = st.button("No!")
    
    if yes_button_clicked:
        st.session_state.feedback_submitted = True
        st.session_state.enjoyed_coffee = True
        
    elif no_button_clicked:
        st.session_state.feedback_submitted = True
        st.session_state.enjoyed_coffee = False
        
    if st.session_state.feedback_submitted:
        if st.session_state.enjoyed_coffee:
            name = st.text_input("Name")
            st.write(f"Thank you {name}, Glad you enjoyed it!")
        else:
            st.write("We're sorry to hear that. We'll strive to improve.")
            st.write("We value your feedback. Please tell us more about your experience.")
            name = st.text_input("Name")
            message = st.text_area("Feedback")
            st.write(f"Thank you {name} for your feedback: {message}")

if page == 'Submit Feedback':
    st.header('Submit Feedback')
    
    st.write("Your input helps us brew a better experience. Please share your thoughts about our coffee and service.")
    
    coffee_rating = st.slider("On a scale of 1 to 10, how would you rate our coffee today?", 1, 10, 5)

    service_type = st.selectbox("What type of service did you experience?", ["In-store", "Online", "Phone"])

    st.write(f"You rated our coffee a {coffee_rating} and experienced {service_type} service.")

    feedback_input()
    
````
