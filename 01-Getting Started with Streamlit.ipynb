# 1) Instroduction

**What is Streamlit?**

Streamlit is an open-source Python library that lets you easily create and share web applications. It's designed to be simple and intuitive, allowing you to prototype and deploy applications quickly without writing extensive HTML, CSS, or JavaScript code.

We're focusing on Streamlit instead of other libraries like **Gradio** or **PyWebIO** because Streamlit offers several advantages that make it an excellent choice for building interactive web applications. Here are a few key reasons:

**Easy to use:** Streamlit provides a simple and intuitive API that allows you to create interactive applications with just a few lines of code. You don't need to be an expert in web development to get started with Streamlit.

**Fast prototyping:** Streamlit makes it easy to iterate quickly and experiment with different ideas. You can see the results of your changes in real-time without restarting your application.

**Data-focused:** Streamlit was initially designed for data science and machine learning applications. It provides built-in support for popular data science libraries like Pandas, NumPy, and Matplotlib, making it easy to work with data and visualize results.

**Integration with AI tools:** Streamlit seamlessly integrates with popular AI tools like OpenAI's chat API, TensorFlow, PyTorch, and scikit-learn, allowing you to integrate machine learning models into your applications.

**Portfolio-ready Projects:** A direct link to your app can help others easily see and test your work, compared to viewing raw code on GitHub.

**Large Network of Users:** Compared to similar libraries, Streamlit has a lot of users, which means it is easier to get support when needed. It also makes Streamlit a more in-demand skill for companies that create in-house dashboards.

# 3) Text Elements

In Streamlit, building an app involves using various components, broadly classified into Elements and Widgets.

* **Elements** are the core components used for displaying information and content in your Streamlit app. They include text elements (like st.title(), st.write()), data elements (for displaying tables and charts), and more specialized types like chat elements. These elements are primarily static because they're used for presenting data, information, or visuals in an app. They form the backbone of an app's structure and content layout.

* **Widgets** are interactive components designed for user engagement and input within an app. They include tools like buttons, sliders, text input fields, and dropdown boxes. Widgets are dynamic, responding to user interactions and enabling real-time updates and functionalities. They play an important role in making apps interactive, allowing them to respond to user inputs and preferences.

**Basic Text Elements**

We got a preview of our first elements, st.title() and st.write, on the previous screen. These functions are part of Streamlit's text elements, which we can use to display various text types. Here are a few of the most useful text elements we'll use throughout the course:

* **st.title()**: Displays a title on the web page.

* **st.header()**: Displays headers smaller than a title but larger than standard text. It can help create distinct sections within your app.

* **st.write()**: Displays text on the web page.

# 4) The Button Widget

Now that we can display titles and various text types let's add interactivity to our app. One of the simplest ways to do this is using **buttons**.

In Streamlit, the `st.button()` [function](https://docs.streamlit.io/develop/api-reference/widgets/st.button) is used to create a button. This function typically takes one argument: a string that we want to display on the button. Here's an example:

```python
st.button("Click me!")
```
In Streamlit, the `st.button()` function returns a boolean value. It will return True when the button is clicked and False otherwise. We can use this to perform an action when the button is clicked. Here's an example:

```python
if st.button("Click me!"):
    st.write("You clicked the button.")
```



In this example, we've used an if statement to check whether the button was clicked. If the button was clicked (i.e., `st.button("Click me!")` is True), the text "You clicked the button." will be displayed on the web page.

**Interacting with Functions**

Streamlit allows us to define elements inside functions for more dynamic interactivity. For instance, we can create a function that is executed when a button is clicked:

```python
def display_message():
    st.write("Welcome to our interactive app!")

if st.button("Greet"):
    display_message()
```



This example demonstrates how clicking the "Greet" button calls the `display_message` function, showing a message on the app.

**Assigning a Button to a Variable**

We can also assign a button to a variable to manage its state more cleanly. This approach enhances code readability and is particularly useful in more complex applications.

 For example:

```python
button_clicked = st.button("Click me!")
if button_clicked:
    st.write("You clicked the button.")
```

**A Note on Button Names**

When creating buttons in Streamlit, it's important to use unique labels, especially if our app contains multiple buttons. Streamlit uses the button's label as part of its unique identifier (or key) by default. If two buttons have the same label, Streamlit might not differentiate between them correctly, leading to unexpected behaviors. For example:

```python
if st.button("Button"):
    st.write("Button 1 was clicked")

if st.button("Button"):
    st.write("Button 2 was clicked")
```

In this snippet, both buttons share the same text ("`Button`"), and when we try to run this app, Streamlit alerts us with a "Duplicate Widget ID" error.

If we want to use the same label on multiple buttons, we can use the optional 'key' parameter to avoid a "Duplicate Widget ID" error. Here's how using "keys" could fix the example from above:

```python
if st.button("Button", key="button1"):
    st.write("Button 1 was clicked")

if st.button("Button", key="button2"):
    st.write("Button 2 was clicked")
```

# 5) Sliders and Dropdown Boxes

In addition to buttons, Streamlit has several other interactive elements. Let's look at sliders and dropdown boxes, which allow users to customize their inputs.

**Sliders**

Sliders in Streamlit allow users to select a value (or range of values) from a continuous or discrete set.

Here's an example using the `st.slider()` [function](https://docs.streamlit.io/develop/api-reference/widgets/st.slider) to create a slider:

```python
st.slider("How many minutes do you code per day?", 0, 100, 50)
```

In this example, a slider will be displayed on the web page with the label "How many minutes do you code per day?". The slider allows users to select a value between 0 and 100, with the initial value set to 50. Let's look at the parameters in our slider example:

* **Label**: 'How many minutes do you code per day?' is the label displayed above the slider.

* **Minimum and Maximum Values (0 and 100)**: These define the range of values the user can select from.

* **Default Value (50)**: This is the initial value when the app loads.

An additional important parameter is:

* **Step**: This defines the increments between values. For instance, a step of 10 in a range of 0-100 allows selections at 0, 10, 20, etc. Adjusting the step size can make the slider more sensitive (smaller steps) or less sensitive (larger steps) to changes.

Like the `st.button()` function, the `st.slider()` function also returns a value. In this case, it returns the current value selected on the slider. We can use this value to perform an action. For example, we assign the slider to a variable called value and use a Python f-string to display it:

```python
value = st.slider("How many minutes do you code per day?", 0, 100, 50)
st.write(f"You selected {value}.")
```

**Dropdown Boxes**

Dropdown boxes, also known as select boxes, are another common interactive element in web applications. They allow users to choose one option from a list of options. The `st.selectbox()` [function](https://docs.streamlit.io/develop/api-reference/widgets/st.selectbox) is used to create a dropdown box and takes two arguments:

1. A **string** for the select box label

1. A **list** of options

Here's an example using the `st.selectbox()` function:

```python
programming_language = st.selectbox("Select a programming language", ["Python", "R", "C++"])
st.write(f"You selected {programming_language}.")
```

* **Label**: 'Select a programming language' is the text displayed above the dropdown box.

* **Options List**: ["Python", "R", "C++"] is the list of options from which the user can choose.

# 6) Text Input and Text Area

**Text Input**

The text input box in Streamlit allows users to enter a **single line of text**. This can be useful for entering simple data such as names, email addresses, or search queries. The `st.text_input` function takes two arguments: 

* label that is displayed above the input box

* An optional placeholder or initial value.

Here's an example of using the `st.text_input()` function to create a text input box:

```python
name = st.text_input("Enter your name")
st.write(f"Hello, {name}!")
```



**Text Area**

The text area box in Streamlit is similar to the text input box, but it allows users to **enter multiple lines of text**. This can be useful for entering longer messages, comments, or descriptions. The `st.text_area` function also takes a label argument. Additionally, we can specify parameters such as height to adjust the size of the input box.

Here's an example of using the `st.text_area()` function to create a text area box:

```python
message = st.text_area("Enter your message")
st.write(f"You entered: {message}")
```



Both `st.text_input()` and `st.text_area()` functions return the current text entered in the box, which we can use to perform an action or display on the web page.

# 7) Additional Input Widgets

The elements we've seen so far form the backbone of most Streamlit applications, facilitating user interaction and data input. However, Streamlit offers many more components to enhance the functionality and aesthetics of our applications.

We'll briefly introduce some of these additional Streamlit elements on this screen, including radio buttons and checkboxes.

## Radio Buttons

Radio buttons are used for **mutually exclusive choices**, meaning the user can select only one option from the group. This makes them particularly useful in scenarios where we need the user to choose a single option from a predefined list.

The `st.radio` function creates a group of radio buttons. We can provide the options as a tuple or a list, each element representing a different radio button. For instance:

```python
option = st.radio(
    "Which option do you prefer?",
    ('Option 1', 'Option 2', 'Option 3'))  # A tuple of options
```

Or using a list:

```python
option = st.radio(
    "Which option do you prefer?",
    ['Option 1', 'Option 2', 'Option 3'])  # A list of options
```



## Checkboxes

Checkboxes in Streamlit allow users to make binary choices, such as **toggling a feature on or off**. They are also used for selecting **multiple items** from a list, offering flexibility that is not available with radio buttons.

However, unlike radio buttons, Streamlit checkboxes don't directly support a list or tuple parameter. Each checkbox is created individually:

```python
agree = st.checkbox("I agree to the terms and conditions.")
```



## Images

Adding images to our app can enhance the visual appeal and convey information effectively. They can be used for various purposes, from displaying data visualizations to adding logos or illustrations.

Streamlit makes it straightforward to incorporate images into your applications using the `st.image` function. This function allows you to display images from a variety of sources, including local files, URLs, or even Python libraries like PIL or OpenCV. (Note that in the Dataquest code editor we will need to use URLs to display images.)

Here's an example of displaying an image from a URL:

```python
st.image("https://example.com/path/to/image.jpg")
```

Alternatively, you can display an image stored locally:

```python
st.image("dataquest_rocket.png")
```

In these examples, the st.image function is used to display the specified images. You can also customize the display by adjusting parameters such as width, use of columns, and image captioning. This functionality allows you to tailor the presentation of images to fit the design and layout of your Streamlit application.
